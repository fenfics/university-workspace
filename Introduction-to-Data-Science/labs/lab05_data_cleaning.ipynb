**3. ให้นำข้อมูลจาก data_clean_bfclean.csv คอลัมน์ SQ_M มาจัดการ**

โดยที่
- ข้อมูลที่มีค่าน้อยกว่า 100 ถือว่าเป็นข้อมูลที่ผิดพลาด ให้เปลี่ยนเป็น Null
- ข้อมูลที่ไม่ใช่ตัวเลขถือเป็นข้อมูลที่ผิดพลาด ให้เปลี่ยนเป็น Null
- ทำคอลัมน์ให้เป็น float64

```python
data_clean = pd.read_csv('data_clean_bfclean.csv')
```

```python
data_clean.info()
```

```python
data_clean.head()
```

```python
def fn(x):
    try:
        x = float(x)
        if x < 100:
            return None
        else:
            return x
    except:
        return None
        
data_clean['SQ_M'] = data_clean['SQ_M'].apply(fn).astype('float64',errors= 'ignore')
```
