```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

df = pd.read_excel('./data/reptile.xlsx')

OE = OrdinalEncoder(categories=[['Male', 'Female']])
df['gender'] = OE.fit_transform(df[['gender']])

scaler = StandardScaler()
df2 = pd.DataFrame()
df2 = df[['temperature', 'gender']]

df3 = pd.DataFrame()
df3[['temperature', 'gender']] = scaler.fit_transform(df2)

att = df3
label = df['water']

att_train, att_test, label_train, label_test = train_test_split(
    att, label, train_size=0.8, random_state=62  
)

model = KNeighborsClassifier(n_neighbors=3)  
model.fit(att_train, label_train)

label_predict = model.predict(att_test)

print(accuracy_score(label_test, label_predict))

model.predict(scaler.transform([[25, 0]]))
```
