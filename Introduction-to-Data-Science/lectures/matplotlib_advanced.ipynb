```python
x = 'Mon Tue Wed Thu Fri'.split()
y = [50,70,35,90,40]
z = [20,54,25,48,25]
```

```python
plt.figure(figsize = [4,3])
plt.barh(x,y)
plt.barh(x,z)
plt.show()
```

```python
plt.figure(figsize = [4,3])
new_x = np.arange(1,len(x)+1)

bw = 0.2
plt.bar(new_x,y,width=bw,label='M')
plt.bar(new_x + bw,z,width=bw,label='W')

plt.xticks(new_x+bw/2,x)
plt.legend(loc='best')
plt.show()
###highlight work flow that u need to know
```
