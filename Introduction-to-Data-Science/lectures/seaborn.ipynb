```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('tips')
df.head()
```

```python
plt.figure(figsize= [4,3])
sns.countplot(data = df, x = 'day',width=0.5,hue='sex',palette='dark:r') 
###u can use shift+tap for looking more command 
plt.title('DAY N SEX')
plt.show()
```

```python
plt.figure(figsize= [4,3])
sns.histplot(data = df, x = 'tip',bins=10,kde=True) 
plt.title('tip')
plt.show()
```

```python
plt.figure(figsize= [4,3])
sns.boxplot(data = df, y = 'tip',x='day') 
plt.title('tip')
plt.show()
```

```python
plt.figure(figsize= [4,3])
sns.violinplot(data = df, y = 'tip',x='day',hue='sex',split=True) 
plt.title('tip')
plt.show()
```

```python
plt.figure(figsize= [4,3])
sns.stripplot(data = df, y = 'tip',x='day',jitter=0.2,hue='sex') 
plt.show()
```

```python
plt.figure(figsize= [10,8])
sns.swarmplot(data = df, y = 'tip',x='day',hue='sex') 
###if it show error its mean your figsize is too short!!!
plt.show()
```

```python
plt.figure(figsize= [4,3])
sns.lineplot(data = df, y = 'tip',x='total_bill',marker='o') 
plt.show()
```

```python
plt.figure(figsize= [4,3])
sns.barplot(data = df, y = 'tip',palette='pastel',hue='day',x='day') 
plt.show()
```

```python
df.select_dtypes(include=['int64','float64']).corr()
```

```python
plt.figure(figsize = [4,3])
sns.heatmap(df.select_dtypes(include=['int64','float64']).corr(),cmap='cool',lw=2)
plt.show()
```

```python
sns.jointplot(data = df , x='tip',y='total_bill',kind='hex',marginal_kws={'color':'red','bins':20,'kde':True},joint_kws={'color':'purple'})
```

```python
g = sns.JointGrid(data = df 
             , x='tip'
             ,y='total_bill')
g.plot_joint(sns.regplot,color='g')
g.plot_marginals(sns.boxplot,color='r')
g.fig.set_size_inches(6,6)
```

```python
sns.pairplot(df)
```

```python
sns.pairplot(df,hue='day')
```

```python
g = sns.PairGrid(df)
g.map_diag(sns.histplot)
g.map_lower(sns.lineplot,marker='s',color='b')
g.map_upper(sns.scatterplot,color='r')
g.fig.set_size_inches(5,5)
```

```python
g = sns.FacetGrid(data = df ,col='day',row='sex')
g.map(sns.scatterplot,'total_bill','tip')
```
