```python
dic = {"day":['Sun','Mon','Tue','Wed','Thu','Fri','Sat'],"temp":[24,25,29,35,23,20,23]}
df = pd.DataFrame(dic)
df
```

```python
plt.figure(figsize=[7,5])
x = df['day']
y = df['temp']
plt.plot(x,y,ls='-.',lw='2',marker='*', c = '#FFC7F5',ms='20',mfc='#FFFCAD',mec='#C7EAFF')
plt.show()
```

```python
plt.figure(figsize=[7,5])
plt.plot(x,y,'o--r')
plt.title('Temp in class 1/22/2026')
plt.xlabel('Day')
plt.ylabel('Temp')
plt.xticks(rotation=90)
plt.yticks([20,22,24,26,28,30,32,34,36])
plt.show()
```

```python
dic = {"day":['Sun','Mon','Tue','Wed','Thu','Fri','Sat'],"temp":[24,25,29,35,23,20,23],"temp2":[25,29,30,38,21,33,24]}
df = pd.DataFrame(dic)
x = df['day']
y = df['temp']
z = df['temp2']
```

```python
plt.plot(x,y,'s-.r',label='Temp1')
plt.plot(x,z,'o:g',label='Temp2')
plt.legend(loc='best')
plt.figure(figsize = [5,4])
plt.show()
```

```python
fig,axes = plt.subplots(nrows = 2 , ncols = 2)
axes[0,1].plot(x,y)
axes[1,0].set_title('week1')
fig.tight_layout()
plt.show()
```

```python
fig,axes = plt.subplots(nrows = 2 , figsize=[5,4])
axes[0].plot(x,y)
axes[1].set_title('week1')
fig.tight_layout()
plt.show()
```
